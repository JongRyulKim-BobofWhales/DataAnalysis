```python
## !!!!!!!!!!!!!!!!!!!!!! 필독  !!!!!!!!!!!!!!!!!!!!!!##
##================= line 6 ~ line 10 확인 =============##
##!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!##

########################################################
## 과제 3의 시각화 작업 시 한글 처리 설정 Start        #
##-----------------------------------------------------#
## 한글 처리를 위한 matplotlib 설정 (1) 부분만 실행 후 #
##  ==> 메뉴 : [런타임 ==> '세션 다시 시작'] 실행 후   #
##  ==> 한글 처리를 위한 matplotlib 설정 (2) 부터 실행 #
########################################################
##-----------------------------------------------------#
## step 1. 한글 처리를 위한 matplotlib 설정 (1)        #
##-----------------------------------------------------#
!sudo apt-get install -y fonts-nanum
!sudo fc-cache –fv
!rm ~/.cache/matplotlib -rf
##-----------------------------------------------------#
## step 2. 메뉴 : [런타임 ==> '세션 다시 시작'] 실행   #
##-----------------------------------------------------#
## step 3. 한글 처리를 위한 matplotlib 설정 (2) 이후 실행 #
##-----------------------------------------------------#
## 한글 처리를 위한 matplotlib 설정 (2)                #
# import matplotlib.pyplot as plt                      #
# plt.rc('font', family='NanumBarunGothic')            #
##-----------------------------------------------------#
## 과제 3. 시각화 작업 한글 처리 설정 End              #
########################################################


##*****************************************************#
## 과제 1-2. 가격 데이터 수집                         *#
##*****************************************************#
import requests
import json
import os

url = 'https://apis.data.go.kr/B552845/perDay/price'

service_key = '812349bfe35559ef8acf4813c5369c681146a86f86b1f2f95fd64b0bffaf6078'

price_data = []

for year in range(2015, 2025):
    params = {
        'serviceKey': service_key,
        'pageNo': 1,
        'numOfRows': 200,
        'returnType': 'JSON',
        'cond[exmn_ymd::GTE]': '{}0101'.format(year),
        'cond[exmn_ymd::LTE]': '{}1231'.format(year),
        'cond[se_cd::EQ]': '02',
        'cond[ctgry_cd::EQ]': '200',
        'cond[item_cd::EQ]': '211',
        'cond[grd_cd::EQ]': '04',
        'cond[sgg_cd::EQ]': '1101',
        'cond[mrkt_cd::EQ]': '0110211',
        'cond[vrty_cd::EQ]': '02'
    }

    try:
        response = requests.get(url, params=params, timeout=5)
        response.raise_for_status()

        data = response.json()

        print('API 응답:', data)

        items = data['response']['body']['items']['item']

        for item in items:
            price_data.append(item)

        print('{}년 수집 건수: {}건'.format(year, len(items)))

    except requests.exceptions.RequestException as e:
        print(f'API 호출 실패: {e}')

    except json.JSONDecodeError as e:
        print(f'JSON 파싱 실패: {e}')

## 데이터 JSON 파일 저장
if len(price_data) > 0:

    file_path = '/content/가락시장_고랭지배추_가격.json'

    try:
        with open(file_path, 'w', encoding='utf-8') as file:
            json.dump(price_data, file, ensure_ascii=False, indent=4)

        if os.path.exists(file_path):
            print('전체 수집 건수: {}건'.format(len(price_data)))

        else:
            print('?? JSON 파일 생성 오류')

    except Exception as e:
        print('?? JSON 파일 저장 오류:', e)

else:
    print('수집된 데이터가 없습니다.')
```

```python
##*****************************************************#
## 과제 2-1. JSON 타입 DataFrame으로 변환             *#
##*****************************************************#
import pandas as pd

file_path = '/content/가락시장_고랭지배추_가격.json'

with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

print('전체 데이터 건수:', len(data))

df = pd.DataFrame(data)

print('\nDataFrame 변환 완료')

print('\n[데이터 기본 정보]')
df.info()
```

```python
##*****************************************************#
## 과제 2-2. DataFrame에 month, season 컬럼 추가      *#
##*****************************************************#
df['exmn_ymd'] = pd.to_datetime(df['exmn_ymd'], format='%Y%m%d')

df['year'] = df['exmn_ymd'].dt.year

df['month'] = df['exmn_ymd'].dt.month

df['season'] = ''

for i in range(len(df)):
    if df.loc[i, 'month'] in [3, 4, 5]:
        df.loc[i, 'season'] = '봄'
    elif df.loc[i, 'month'] in [6, 7, 8]:
        df.loc[i, 'season'] = '여름'
    elif df.loc[i, 'month'] in [9, 10, 11]:
        df.loc[i, 'season'] = '가을'
    else:
        df.loc[i, 'season'] = '겨울'

df = df.drop('month', axis=1)

print('[연도, 계절 컬럼 추가 완료]')
print(df.head(5))
```

```python
##*****************************************************#
## 과제 3-1. Data 시각화_연도별 선 그래프             *#
##*****************************************************#
df['exmn_dd_prc'] = pd.to_numeric(df['exmn_dd_prc'], errors='coerce')

year_price = df.groupby('year')['exmn_dd_prc'].mean()

year_price.plot(kind='line', marker='o')

plt.title('연도별 배추 평균 가격(8608)')
plt.xlabel('연도')
plt.ylabel('평균 가격')

plt.grid(True)
plt.show()
```

```python
##*****************************************************#
## 과제 3-2. Data 시각화_계절별 막대 그래프           *#
##*****************************************************#
season_price = df.groupby('season')['exmn_dd_prc'].mean()

season_price = season_price.reindex(['봄', '여름', '가을', '겨울'])

bar = season_price.plot(kind='bar')

for i, value in enumerate(season_price):
    bar.text(i, value, round(value, 2), ha='center', va='bottom')

plt.title('계절별 배추 평균 가격(8608)')
plt.xlabel('계절')
plt.ylabel('평균 가격')

plt.show()
```
